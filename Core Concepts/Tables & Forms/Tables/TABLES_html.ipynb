# HTML TABLES
#### An HTML table is defined using the table tag, which organizes structured data into a grid layout. Within this container, rows are built using the tr (table row) tag, while the actual cells are created with th for bold, centered header text and td for standard table data
------

```html
<td> --- Table Data
<tr> --- Table Row
<th> --- Table Header
```
--------

````python
%%html

<table>
  <tr>
    <th>Product</th>
    <th>Quantity</th>
  </tr>
  <tr>
    <td>Apples</td>
    <td>12</td>
  </tr>
  <tr>
    <td>Oranges</td>
    <td>5</td>
  </tr>
</table>
````

````python
%%html

<table>
    <tr>
        <th>Rank</th>
        <th>Animal</th>
        <th>Binomial Name</th>
        <th>Average Mass</th>
    </tr>
    <tr>
        <td>1</td>
        <td>Ostrich</td>
        <td>Struthio camelus</td>
        <td>104</td>
    </tr>
        <tr>
        <td>2</td>
        <td>Lion</td>
        <td>dasdsad camelus</td>
        <td>204</td>
    </tr>
        <tr>
        <td>3</td>
        <td>Rabbit</td>
        <td>Sdasdasdo camelus</td>
        <td>124</td>
    </tr>
</table>
````

# Using thead and tbody
```html
<thead></thead>
<tbody></tbody>
```

````python
%%html

<table>
    <thead>
        <tr>
            <th>Rank</th>
            <th>Animal</th>
            <th>Binomial Name</th>
            <th>Average Mass</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td>1</td>
            <td>Ostrich</td>
            <td>Struthio camelus</td>
            <td>104</td>
        </tr>
            <tr>
            <td>2</td>
            <td>Lion</td>
            <td>dasdsad camelus</td>
            <td>204</td>
        </tr>
            <tr>
            <td>3</td>
            <td>Rabbit</td>
            <td>Sdasdasdo camelus</td>
            <td>124</td>
        </tr>
    <tbody>
</table>
````
