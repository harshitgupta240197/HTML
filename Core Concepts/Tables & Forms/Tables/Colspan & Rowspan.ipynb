# colspan & rowspan
#### In HTML tables, colspan and rowspan are attributes used on td (table data) or th (table header) tags to merge adjacent cells into a single larger cell. They function exactly like the "Merge Cells" feature in spreadsheet software
colspan (Column Span): Merges cells horizontally across multiple columns.<br>
rowspan (Row Span): Merges cells vertically down across multiple

```python
%%html

<table border="1">
  <tr>
    <!-- This header spans across 2 columns -->
    <th colspan="2">Contact Information</th>
  </tr>
  <tr>
    <td>Email</td>
    <td>Phone</td>
  </tr>
</table>
```

```python
%%html

<table border="1">
  <tr>
    <!-- This cell drops down into row 2 -->
    <td rowspan="2">User Details</td>
    <td>Name: Alice</td>
  </tr>
  <tr>
    <!-- Do NOT put a first cell here; the space is already taken -->
    <td>Age: 25</td>
  </tr>
</table>
```
