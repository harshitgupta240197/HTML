# Form Validation
#### HTML form validation is a built-in browser mechanism that checks user data against predefined constraints before allowing form submission. It improves data integrity, provides immediate user feedback, and reduces unnecessary server loads
- client side validation
- sevrer side validation
--------
---------

## Core Validation Attributes
- `required`: Marks the field as mandatory.
- `minlength` / `maxlength`: Restricts text character counts.
- `min` / `max`: Restricts numerical or date ranges.
- `pattern`: Accepts a Regular Expression (Regex) that the input must match.
- `type`: Dictates semantic data structure (e.g., `email`, `url`, `number`).
_________
_______

```python
%%html

<form action="/submit-data" method="POST">
  <!-- Required text field with length boundaries -->
  <label for="username">Username:</label>
  <input type="text" id="username" name="username" minlength="3" maxlength="15" required>
<br>
<br>
  <!-- Built-in semantic email validation -->
  <label for="email">Email Address:</label>
  <input type="email" id="email" name="email" required>
<br>
<br>
  <!-- Number validation with range limits -->
  <label for="age">Age (18-99):</label>
  <input type="number" id="age" name="age" min="18" max="99">
<br>
<br>
  <!-- Regular Expression pattern validation (Format: 123-456-7890) -->
  <label for="phone">Phone (Format: 123-456-7890):</label>
  <input type="tel" id="phone" name="phone" pattern="[0-9]{3}-[0-9]{3}-[0-9]{4}" title="Please match the requested format: 123-456-7890">
<br>
<br>
  <button type="submit">Submit</button>
</form>
```
