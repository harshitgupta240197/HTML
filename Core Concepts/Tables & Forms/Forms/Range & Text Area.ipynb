# Range and Text Area in html
#### In HTML, ```<input type="range">``` creates a slider control for selecting imprecise numeric values
#### ```<textarea>``` creates a multi-line text input field for long paragraphs, comments, or reviews

````python
%%html

<!-- Example of a volume slider from 0 to 10 starting at 5 -->
<label for="volume">Volume:</label>
<input type="range" id="volume" name="volume" min="0" max="10" step="1" value="5">
````

````python
%%html

<!-- Example of a feedback text box -->
<label for="feedback">Your Review:</label><br><br>
<textarea id="feedback" name="feedback" rows="4" cols="50" placeholder="Type your comment here..."></textarea>
````
