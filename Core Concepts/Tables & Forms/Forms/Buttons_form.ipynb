# Buttons in html forms
#### In HTML forms, buttons are primarily created using the ```<button>``` element or the ```<input>``` element. Modern web development heavily favors the ```<button>``` tag because it is a paired tag that can contain nested HTML content like text, formatting tags, icons, or images, whereas ```<input>``` is a self-closing tag restricted to plain text via its value attribute

````python
%%html

<form action="/submit-data" method="POST">
  <!-- Inputs here -->
  
  <!-- Submit type (default) -->
  <button type="submit">Send Form</button>
  
  <!-- Generic type for JS actions -->
  <button type="button" onclick="alert('Hello!')">Click Me</button>
</form>
````
