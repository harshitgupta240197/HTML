# Name attribute in html forms
#### In an HTML form, the name attribute is used as the identifier for form data after it is submitted to a server, acting as the "key" in the sent key-value pairs. Without a name attribute, an input element's value cannot be sent to the server during a traditional form submission

```python
%%html

<form action="/submit-data" method="POST">
  <!-- The server reads this data using the key "username" -->
  <label for="user">Username:</label>
  <input type="text" id="user" name="username">

  <button type="submit">Submit</button>
</form>
```
