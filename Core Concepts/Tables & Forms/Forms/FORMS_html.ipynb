# FORMs in html
#### An HTML form is used to collect user input and send it to a server for processing. It acts as a container for interactive controls like text fields, checkboxes, and buttons

````python
%%html 

<form action="/submit-form-endpoint" method="POST">
  <!-- Text Input -->
  <label for="username">Username:</label>
  <input type="text" id="username" name="username" placeholder="Enter username" required>

  <!-- Email Input -->
  <label for="email">Email Address:</label>
  <input type="email" id="email" name="email" required>

  <!-- Dropdown Menu -->
  <label for="topic">Choose a Topic:</label>
  <select id="topic" name="topic">
    <option value="support">Technical Support</option>
    <option value="sales">Sales</option>
  </select>

  <!-- Submit Button -->
  <button type="submit">Submit Data</button>
</form>
````

# INPUT TAG
#### The HTML input tag is used to create interactive controls within web forms to collect data from users. It is an empty (self-closing) element, meaning it contains attributes only and does not require a closing input tag

````python
%%html

<form action='/tacos'>
    <input>
</form>
````

## Type attribute inside the INPUT tag

````python
%%html

<form action='/tacos'>
    <input type='text'>
    <input type='password'>
    <input type='color'>
    <input type='date'>
    <input type='number'>
</form>
````

## PLACEHOLDER text attribute in input tag

````python
%%html

<form action='/tacos'>
    <input type='text' placeholder='type here'>
    <input type='password' placeholder='Your Password'>
    <input type='color'>
    <input type='date'>
    <input type='number' placeholder = 0>
</form>
````

# LABEL in html forms
#### The ```<label>``` HTML element represents a caption for an item in a user interface, providing a clickable, accessible text descriptor for form controls like inputs, textareas, checkboxes, and dropdowns. It bridges the gap between what a human reads visually and what a web browser or screen reader interprets programmatically

````python
%%html

<label for="username">Username:</label>
<input type="text" id="username" name="username">
````
