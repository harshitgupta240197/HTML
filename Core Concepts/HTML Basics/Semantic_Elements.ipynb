# Semantic Elements
#### Semantic elements in HTML are tags that clearly describe their meaning and purpose to both the browser and the developer. Instead of just defining how content should look, they define what the content is.
-----

## Main / Nav
```html
<html>

    <head>
        <title>page title</title>
    </head>
       
        <body>
            <nav>
                <ul>
                    <li>content</li>
                    <li>content</li>
                    <li>content</li>
                </ul>
            </nav>

            <main>
                Content
            </main>
            <footer>
                This is your footer
            </footer>
        </body>
        
</html>
```
