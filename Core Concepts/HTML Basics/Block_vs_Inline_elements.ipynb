# BLOCK vs INLINE elements
#### In HTML, the primary difference is that block-level elements always start on a new line and take up the full available width of their parent container, whereas inline elements stay on the same line and only take up as much space as their content requires
-----------------
# DIV
The div tag is a generic container element used to group other HTML elements together. It has no semantic meaning or visual effect on its own, but it is crucial for structuring web pages, applying CSS styles, and manipulating content with JavaScript
```html
<div></div>
```
---------------

# SPAN
#### The span tag in HTML is a generic, non-semantic inline container used to group and style specific sections of text or phrasing content. It has no structural or visual impact on its own; it is designed to target elements via CSS or JavaScript using class or id attributes
```html
<span></span>
```
