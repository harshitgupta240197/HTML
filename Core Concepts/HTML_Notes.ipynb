## BOLD TAG
```html
<b>your_text</b>
```

## PARAGRAPH TAG
```html
<p>your_text</p>
```

## HEADING TAGS
```html
<h1>your_text</h1>
<h2>your_text</h2>
<h3>your_text</h3>
<h4>your_text</h4>
<h5>your_text</h5>
<h6>your_text</h6>
```
##### We should always only have one H1 on a page
----------------

# HTML FORMAT
```html
<!DOCTYPE html>
<html>
    <head>
        <title>page title</title>
    </head>
    <body>
        content to show on the page
    </body>
</html>
```
#### The shortcut to this boilerplate is ! + Tab
----------------

## HEAD element is the document metadata element
#### It basically contains everything that we do not want to display on the page
1. info on the webpage
2. scripts that we want to run
3. CSS and styling
-----------

# LISTS
## UNORDERED LIST
```html
<ul>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
</ul>
```

## ORDERED LIST
```html
<ol>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
    <li>content</li>
</ol>
```

## NESTED LISTS
```html
<ul>
    <li>Category</li>
    <li>Category</li>
    <li>Subcategory
        <ol>
            <li>This is neested list</li>
            <li>This is neested list</li>
            <li>This is neested list</li>
            <li>This is neested list</li>
        </ol>
    </li>
</ul>
```
-----------
